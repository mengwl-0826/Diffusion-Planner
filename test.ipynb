```python
import torch

# 假设ego_future包含3个未来时间步的[x, y, θ]
ego_future = torch.tensor([
    [10.0, 20.0, 0.7854],  # θ=π/4 (45度)
    [15.0, 25.0, 1.5708],  # θ=π/2 (90度)
    [20.0, 30.0, 3.1416],  # θ=π (180度)
])

try_stack = torch.stack(
                    [ego_future[..., 2], ego_future[..., 2]], dim=-1
                )

# 提取朝向角并转换为方向向量
ego_future = torch.cat(
            [
                ego_future[..., :2],
                try_stack,
            ],
            dim=-1,
            )
print("原始角度(弧度):")
print(ego_future[..., 2].size(), ego_future[..., 2])

print("try_stack:")
print(try_stack.size(), try_stack)

print("after ego_future:")
print(ego_future.size(), ego_future)
```

```python
a = torch.randn(3, 2)  
b = torch.randn(3, 4)  
acatb = torch.cat([a, b], dim=-1)  # 沿第-1维拼接，其他维度形状相同，合法

print('a:', a.size(), a)
print('b:', b.size(), b)
print('acatb:', acatb.size(), acatb)


a = torch.randn(1, 3)  
b = torch.randn(1, 3)  
astackb = torch.stack([a,b], dim=0)    # 所有输入张量形状必须完全相同, 在哪个维度stack就增加一个维度
print('a:', a.size(), a)
print('b:', b.size(), b)
print('astackb:', astackb.size(), astackb)  #astackb: torch.Size([2, 1, 3])
astackb = torch.stack([a,b], dim=-1) 
print('astackb:', astackb.size(), astackb)  #astackb: torch.Size([1, 3, 2])
```

```python
import numpy as np
import torch
ego_current = np.array([1, 2, 3])   # 形状：(3, )表示一维数组
result = ego_current[:, None]  # 形状：(3, 1)
# 等价于：
result = ego_current.reshape(-1, 1)  # NumPy 或 PyTorch 均可使用
# 或使用 unsqueeze：
result = torch.from_numpy(ego_current).unsqueeze(1)  # PyTorch

print(result.size(), result)
```

```python
import torch

# 创建一个形状为 [2, 3] 的张量
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
], dtype=torch.float32)

# 沿第1维求平均（即对每一行的元素求平均）
result = torch.mean(x, dim=1)

print("原始张量 x:\n", x)
print("形状:", x.shape)  # 输出: torch.Size([2, 3])
print("\n处理后的张量:\n", result)
print("形状:", result.shape)  # 输出: torch.Size([2])
# 结果: tensor([2., 5.])
```
